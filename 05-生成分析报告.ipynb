```python
totals={}
try:
    with open('data/ridership.txt','r',encoding='utf-8') as f:
        for line in f:
            line=line.strip()
            if not line:
                continue
            parts=line.replace('，',',').split(',')
            if len(parts)<2:
                print(f"错误:数据残缺,已跳过：{line}")
                continue
            try:
                totals[parts[0]]=int(parts[1])
            except ValueError:
                print("数据不是有效数字，跳过")
                continue
    print(totals)
except FileNotFoundError:
    print("错误：未找到data/ridership.txt,请检查data目录")
except Exception as e:
    print(f"错误：{e}")
```

```python
energy={}
try:
    with open('data/energy.csv','r',encoding='utf-8') as f:
        for line in f:
            line=line.strip()
            if not line:
                continue
            parts=line.replace('，',',').split(',')
            line_name=parts[0]
            km=float(parts[1])
            kwh=float(parts[2])
            try:
                energy[line_name]={'km':km,'kwh':kwh}
            except ValueError:
                print(f"数据不是有效数字:{line_name}")
    print(energy)
except FileNotFoundError:
    print("错误，文件data/energy.csv不存在，请检查data目录")
except Exception as e:
    print(f"发现未知错误：{e}")
```

```python
lines=[]
intensity={}
for name in totals:
    ridership=totals[name]
    energy_kwh=energy[name]['kwh']
    try:
        result=energy_kwh/ridership
        intensity[name]=result
        lines.append(f"{name}:{result:.2f}kwh/万人次")
    except ZeroDivisionError:
        print(f"除零错误:{name}")
```

```python
lines.insert(0,"地铁线路能耗与客流分析报告")
widest=max(len(line) for line in lines)
lines.insert(1,"——"*widest)
lines.insert(2,"")
if intensity:
    min_line=min(intensity,key=intensity.get)
    min_val=intensity[min_line]
    lines.append("")
    lines.append(f"结论：单位电耗最低的线路是：{min_line},为{min_val:.2f}kwh/万人次。")
report_content='\n'.join(lines)
try:
    with open('output_report.txt','w',encoding='utf-8') as f:
        f.write(report_content)
        print("报告已成功生成：output_report.txt")
except FileNotFoundError:
    print("错误：未识别到output_report.txt")
except Exception as e:
    print(f"未知错误：{e}")
```
