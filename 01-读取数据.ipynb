```python
print("开始读取客流数据")
try:
    with open('data/ridership.txt','r',encoding='utf-8') as f:
        total_ridership=0
        for line in f:
            stripped_line=line.strip()
            if not line.strip():
                continue
            parts=stripped_line.strip().replace('，',',').split(',')
            if len(parts)<2:
                print(f"发现格式错误行，已跳过：'{stripped_line}'")
                continue
            name=parts[0]
            count=int(parts[1])
            print(f"{name}客流：{count}")
            total_ridership+=count
    print(f"总客流：{total_ridership}")
except FileNotFoundError:
    print("错误：ridership.txt不存在，请检查路径")
except Exception as e:
    print(f"发生未知错误：{e}")
print("程序正常结束")
```
