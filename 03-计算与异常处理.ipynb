```python
ridership={}
try:
    with open('data/ridership.txt','r',encoding='utf-8') as f:
        for line in f:
            line=line.strip()
            if not line:
                continue
            parts=line.replace("，",",").split(',')
            ridership[parts[0]]=int(parts[1])
except FileNotFoundError:
    print("错误：找不到data/ridership.txt,请先建立文件")
except Exception as e:
    print(f"读取客流数据失败：{e}")
energy={}
try:
    with open('data/energy.csv','r',encoding='utf-8') as f:
        for line in f:
            line=line.strip()
            if not line:
                continue
            parts=line.replace('，',',').split(',')
            name=parts[0]
            km=float(parts[1])
            kwh=float(parts[2])
            energy[name]={'km':km,'kwh':kwh}
except FileNotFoundError:
    print("错误：找不到data/energy.csv")
except ValueError as e:
    print(f"能耗数据格式错误：{e}")
print("\n各线路每百万人次电耗：")
for name,total in ridership.items():
    if name not in energy:
        print(f"{name}:缺少能耗数据")
        continue
    kwh=energy[name]['kwh']
    try:
        intensity=kwh/total*10000
        print(f"{name}:{intensity:.2f}kwh/万人次")
    except ZeroDivisionError:
        print(f"{name}:客流量为0，无法计算")
    except Exception as e:
        print(f"{name}:计算异常{e}")
print("\n程序运行结束")
```
