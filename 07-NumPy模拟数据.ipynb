```python
import numpy as np
rng=np.random.default_rng(seed=42)
```

```python
ridership=rng.integers(low=50000,high=150000,size=(3,90))
print("数组形状：",ridership.shape)
```

```python
print("第1条线路前五天客流：",ridership[0,:5])
```

```python
line_names=['1号线','2号线','3号线']
dates = [f'2026-01-{i+1:02d}' for i in range(31)] + \
        [f'2026-02-{i+1:02d}' for i in range(28)] + \
        [f'2026-03-{i+1:02d}' for i in range(31)]
print("日期数量：",len(dates))
```

```python
with open('data/ridership_daily.csv','w',encoding='utf-8')as f:
    f.write('线路,日期,客流量\n')
    for i in range(3):
        line_name=line_names[i]
        for j in range(90):
            f.write(f'{line_name},{dates[j]},{ridership[i,j]}\n')
print("数据已生成:data/ridership_daily.csv")
```

```python
data=np.genfromtxt('data/ridership_daily.csv',
                   delimiter=',',
                   skip_header=1,
                   usecols=2,
                   dtype=int,
                   encoding='utf-8')
data_reshaped=data.reshape(3,90)
```

```python
print("各线路客流统计：")
for i,name in enumerate(line_names):
    line_data=data_reshaped[i]
    total=line_data.sum()
    daily_mean=line_data.mean()
    print(f"{name}:总客流{total}人次,日均{daily_mean:.0f}人次")
```
